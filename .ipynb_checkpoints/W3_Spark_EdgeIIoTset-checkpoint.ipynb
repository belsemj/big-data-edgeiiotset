# Détection d'intrusion IoT sur **Spark**

On reprend le pipeline de détection d'intrusion du W2 (dataset **Edge-IIoTset**),
mais réécrit en **PySpark** (Spark DataFrame + MLlib) au lieu de pandas/scikit-learn.

Idée (MapReduce -> Spark) : le traitement est distribué sur un cluster,
les données sont des *RDD/DataFrame* partitionnés et gardés en mémoire, et les
opérations (transformations + actions) sont optimisées automatiquement.

**Cible :** `Attack_type` (15 classes : trafic normal + 14 attaques).

### Démarrage de la session Spark

```python
import time, glob
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = (SparkSession.builder
         .appName("W3-EdgeIIoTset")
         .master("local[*]")                 # local[*] = tous les coeurs de la machine
         .config("spark.sql.shuffle.partitions", "8")
         .config("spark.driver.memory", "4g")              # augmente si ta machine le permet (ex "6g")
         .config("spark.sql.constraintPropagation.enabled", "false")  # evite l'OOM du planificateur
         .getOrCreate())
spark.sparkContext.setLogLevel("ERROR")
# utile si une session tournait déjà (réglage applicable à chaud)
spark.conf.set("spark.sql.constraintPropagation.enabled", "false")
print("Spark version :", spark.version)
```

### Chargement des données

```python
# Trouve le CSV (local : à côté du notebook ; Kaggle : /kaggle/input)
hits = glob.glob("**/ML-EdgeIIoT-dataset.csv", recursive=True) \
     + glob.glob("/kaggle/input/**/ML-EdgeIIoT-dataset.csv", recursive=True)
CSV_PATH = hits[0]
print("CSV_PATH =", CSV_PATH)

df = spark.read.csv(CSV_PATH, header=True, inferSchema=True)

# IMPORTANT : les noms de colonnes contiennent des points (http.request.method...)
# que Spark interprète comme des champs imbriqués -> on remplace "." par "_"
df = df.toDF(*[c.replace(".", "_") for c in df.columns])

print("Dimensions :", df.count(), "lignes x", len(df.columns), "colonnes")
df.limit(5).toPandas()
```

### Nettoyage

On enlève les colonnes identifiants / à fuite (IP, MAC, payloads, URI...),
les lignes vides et les doublons, exactement comme au W2.

```python
cols_a_supprimer = ["frame_time","ip_src_host","ip_dst_host","arp_src_proto_ipv4",
    "arp_dst_proto_ipv4","http_file_data","http_request_full_uri",
    "icmp_transmit_timestamp","http_request_uri_query","tcp_options",
    "tcp_payload","tcp_srcport","tcp_dstport","udp_port","mqtt_msg"]

df = df.drop(*[c for c in cols_a_supprimer if c in df.columns])
df = df.dropna().dropDuplicates()
print("Après nettoyage :", df.count(), "lignes x", len(df.columns), "colonnes")
```

### Préparation des features (pipeline MLlib)

En Spark, on ne manipule pas des colonnes pandas mais un **pipeline** d'étapes :
- `StringIndexer` encode la cible texte en indices numériques (`label`),
- les colonnes catégorielles sont indexées puis passées en one-hot (`OneHotEncoder`),
- `VectorAssembler` regroupe toutes les features en un seul vecteur,
- `StandardScaler` normalise (comme le `StandardScaler` de sklearn au W2).

> Pour éviter un `OutOfMemoryError` : on **retire** les colonnes texte à forte
> cardinalité (sinon le one-hot crée un vecteur géant) et on utilise
> `withMean=False` dans le scaler (garde les vecteurs creux).

```python
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler, StandardScaler
from pyspark.ml import Pipeline

TARGET = "Attack_type"
feature_cols = [c for c in df.columns if c not in ("Attack_label", "Attack_type")]

# séparer colonnes texte / numériques
cat_cols = [c for c, t in df.dtypes if t == "string" and c in feature_cols]
num_cols = [c for c in feature_cols if c not in cat_cols]
print("Catégorielles :", cat_cols)

# garde-fou : certaines colonnes texte ont des chaînes vides -> "NA" (sinon one-hot plante)
for c in cat_cols:
    df = df.withColumn(c, F.when((F.col(c).isNull()) | (F.trim(F.col(c)) == ""), "NA")
                           .otherwise(F.col(c)))

# On n'encode en one-hot QUE les colonnes a faible cardinalite.
# Les colonnes texte a forte cardinalite (ex. dns.qry.name, mqtt.topic) feraient exploser
# la memoire (vecteur geant -> OutOfMemoryError) : on les retire des features.
MAX_CARD = 40
card = {c: df.select(c).distinct().count() for c in cat_cols}
cat_drop = [c for c in cat_cols if card[c] > MAX_CARD]
cat_cols = [c for c in cat_cols if card[c] <= MAX_CARD]
if cat_drop:
    print("Colonnes texte retirees (trop de valeurs distinctes) :", {c: card[c] for c in cat_drop})
    df = df.drop(*cat_drop)
print("Categorielles encodees :", {c: card[c] for c in cat_cols})

label_idx = StringIndexer(inputCol=TARGET, outputCol="label", handleInvalid="keep")
idxers = [StringIndexer(inputCol=c, outputCol=c + "_idx", handleInvalid="keep") for c in cat_cols]
ohe = (OneHotEncoder(inputCols=[c + "_idx" for c in cat_cols],
                     outputCols=[c + "_ohe" for c in cat_cols]) if cat_cols else None)

asm_inputs = num_cols + ([c + "_ohe" for c in cat_cols] if cat_cols else [])
assembler = VectorAssembler(inputCols=asm_inputs, outputCol="features_raw", handleInvalid="keep")
scaler = StandardScaler(inputCol="features_raw", outputCol="features", withMean=False, withStd=True)  # withMean=False : garde les vecteurs creux

stages = [label_idx] + idxers + ([ohe] if ohe else []) + [assembler, scaler]
prep = Pipeline(stages=stages)
```

### Séparation train / test (60 / 40)

```python
train, test = df.randomSplit([0.6, 0.4], seed=42)

# on ajuste le pipeline de préparation sur le train uniquement (pas de fuite), puis on transforme
prep_model = prep.fit(train)
train_p = prep_model.transform(train).select("features", "label").cache()
test_p  = prep_model.transform(test).select("features", "label").cache()

classes = prep_model.stages[0].labels   # noms des classes dans l'ordre des indices
print("Classes :", classes)
print("train :", train_p.count(), " | test :", test_p.count())
```

### Entraînement des modèles

```python
from pyspark.ml.classification import (LogisticRegression, DecisionTreeClassifier,
                                       RandomForestClassifier)

modeles = {
    "Regression Logistique": LogisticRegression(maxIter=100),
    "Arbre de decision":     DecisionTreeClassifier(seed=42, maxDepth=20),
    "Random Forest":         RandomForestClassifier(numTrees=100, seed=42, maxDepth=20, maxBins=64),
}

modeles_fit, temps = {}, {}
for nom, m in modeles.items():
    t0 = time.time()
    modeles_fit[nom] = m.fit(train_p)
    temps[nom] = time.time() - t0
    print(f"{nom} : entraîné en {temps[nom]:.1f} s")
```

### Évaluation

```python
from pyspark.ml.evaluation import MulticlassClassificationEvaluator
import pandas as pd

metriques = ["accuracy", "weightedPrecision", "weightedRecall", "f1"]
evals = {mt: MulticlassClassificationEvaluator(metricName=mt) for mt in metriques}

lignes = []
preds = {}
for nom, fit in modeles_fit.items():
    p = fit.transform(test_p)
    preds[nom] = p
    lignes.append({
        "Modele": nom,
        "Temps (s)": round(temps[nom], 1),
        "Accuracy":  round(evals["accuracy"].evaluate(p), 4),
        "Precision": round(evals["weightedPrecision"].evaluate(p), 4),
        "Recall":    round(evals["weightedRecall"].evaluate(p), 4),
        "F1":        round(evals["f1"].evaluate(p), 4),
    })

resultats = pd.DataFrame(lignes).set_index("Modele")
resultats
```

### Matrice de confusion du meilleur modèle (F1)

```python
best_nom = resultats["F1"].idxmax()
print("Meilleur modèle :", best_nom)

# on ramène la matrice de confusion côté pandas (petit tableau agrégé)
cm = (preds[best_nom].groupBy("label", "prediction").count().toPandas())
mat = (cm.pivot_table(index="label", columns="prediction", values="count", fill_value=0))

import numpy as np, matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(9, 8))
im = ax.imshow(mat.values, cmap="Blues")
ax.set_xticks(range(mat.shape[1])); ax.set_yticks(range(mat.shape[0]))
ax.set_xticklabels([classes[int(i)] if int(i) < len(classes) else i for i in mat.columns],
                   rotation=90)
ax.set_yticklabels([classes[int(i)] if int(i) < len(classes) else i for i in mat.index])
ax.set_xlabel("Prédiction"); ax.set_ylabel("Vraie classe")
ax.set_title(f"Matrice de confusion - {best_nom}")
fig.colorbar(im); plt.tight_layout(); plt.show()
```
